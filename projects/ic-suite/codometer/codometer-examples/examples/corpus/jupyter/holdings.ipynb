# Holdings

Reads the sample inventory and reports what is on hand.


```python
from dataclasses import dataclass

MINIMUM_ON_HAND = 1
```

```python
@dataclass(frozen=True)
class Holding:
    """How many of one item sit at one location."""

    item_identifier: str
    quantity: int


def on_hand(holdings: list[Holding]) -> int:
    """Return the total quantity held."""
    return sum(holding.quantity for holding in holdings)


print(on_hand([Holding("item-1", 1), Holding("item-2", 2)]))
```

## What the numbers mean

A [holding](https://example.com/holdings) is one item at one location.


```python
on_hand([Holding("item-1", 1)]) >= MINIMUM_ON_HAND
```
